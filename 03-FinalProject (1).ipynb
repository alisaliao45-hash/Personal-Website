## Abstract

This project investigates whether the popularity of U.S. reality dating shows (measured by their viewership and ratings in each year) is associated with dating app activity (measured by their revenue and user counts each year) in the United States from 2019 to 2026. Prior research suggests that both dating shows and dating apps share similar sociocultural sentiments regarding romance and relationships, and we sought to see if this association can also be reflected by the data. We focused on three specific major dating shows, Love Island USA, The Bachelor, and Love Is Blind, and we examined their ratings and viewership along with annual user count and annual revenue for Tinder, Bumble, and Hinge. After some exploratory analysis, we found issues with multicollinearity, autocorrelation, and skewness in our datasets, which we fixed by segregating potentially correlating predictors, incorporating Newey-West Robust Standard Errors, and applying log transformations, respectively. For our main analysis, we applied 12 OLS regression models (combinations of our 6 predictor variables and 2 prediction variables) to test whether dating show popularity could predict dating app user counts and revenue. Our results suggest that there is typically a positive relationship between dating show popularity and dating app popularity, though this was not true for every show, and it often depended on if we measured dating app popularity based on viewership data or ratings data. The Bachelor consistently produced the strongest results. Our findings provide preliminary evidence that dating shows and dating apps may reflect similar social interests and attitudes toward dating. However, our study is limited by a small sample size and the availability of the access of public data. Future research using larger datasets and additional measures of dating behavior would help determine whether this relationship remains consistent over time.

## 1.2. Background and Prior Work

The things that happen in dating shows are not like actual dating. Our project is motivated by the connections we see between dating shows and dating apps. They both seem to be related to the sexualization of women, superficial perceptions of romantic connection, and the perpetuation of unrealistic beauty standards. Although we are unable to run an experiment and establish any causal mechanisms between the two, we sense that the similar sociocultural sentiments underlying both dating shows and dating apps should be reflected by a correlation between dating show viewership/ratings and dating app usage. Beyond the aforementioned studies that look at either dating shows or dating apps separately, we find very little prior empirical research on this topic. However, others have also previously speculated that both dating shows and dating apps (particularly Love Island and Tinder, respectively) collectively contribute to how intimacy has become more "performative" and "transactional" in the modern day.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5)

#### 3.1.2. Relationship Between Viewership and Ratings

Next, we test for any potential relationship between a show's viewership data and its ratings for a given season. Intuitively, we would expect a better show (i.e., with higher ratings) to attract more viewers, and we want to see if the data confirms this intuition. Perhaps more importantly, we need to make sure that there won't be any collinearity between our predictor variables in our eventual model. That is, if we find a significant relationship between a show's annual viewership and ratings, we cannot use both variables in a linear model that predicts dating app usage in that same year.

First, some data wrangling is necessary. We need to combine the viewership and ratings datasets, such that each observation (row) is meant for a single season of a show, and that season's viewership and rating should also be on that row. 

```python
# combine two datasets into one

show_ratings = show_ratings.rename(columns={'show name': 'show_name', 'season': 'season_num'})

season_ratings = show_ratings.groupby(['show_name', 'season_num'])['rating'].mean().reset_index()
season_ratings = season_ratings.rename(columns={'rating': 'average_rating'})

combined = pd.merge(show_viewership, season_ratings, on=['show_name', 'season_num'])
combined = combined.set_index('season_num')
combined = combined.drop(columns='above_2.5mil')

combined
```

Now that all relevant information has been organized into one single dataset, we now make a scatterplot with ratings on the x-axis and viewership on the y-axis. We include OLS best-fit lines with 95% confidence intervals to more directly show the relationship. 

```python
# for all seasons, regardless of which show they belong to

sns.regplot(
    data=combined,
    x='average_rating',
    y='num_viewers (mil)',
    scatter=False,
    ci=95,
    color='black',
    line_kws={'label': 'All shows'}
)
```

```python
# plot a line for shows seperately

sns.lmplot(
    data = combined,
    x = 'average_rating',
    y = 'num_viewers (mil)',
    hue = 'show_name',
    height = 6,
    aspect = 1.3,
    ci = 95
)

plt.title('Average Rating vs. Number of Viewers')
plt.xlabel('Average Rating')
plt.ylabel('Viewers (millions)')

plt.show()
```

This is quite unexpected. For all three shows, as well as when they are all combined, we find that a season's ratings and viewership are actually negatively correlated; a "better" show ends up attracting fewer viewers. One possible explanation for this finding might relate to systematic bias in who rates a show. Perhaps only a select subset of relatively dedicated viewers of a show watch the less popular seasons; since they are dedicated, they may have more favorable opinions towards the show overall, which leads to them leaving higher ratings, even for those less popular seasons. Less dedicated viewers are not present to drag the average rating down. This is somewhat akin to certain artists' most critically acclaimed songs having relatively few streams, while less acclaimed songs have more streams.

Finally, we calculate the Pearson's R coefficient for each correlation, as well as the p-value. However, note that the p-value, no matter its value, wouldn't be that persuasive anyways, given our limited sample size. 

```python
from scipy.stats import pearsonr

results = []

for show, show_data in combined.groupby('show_name'):
    r, p_value = pearsonr(
        show_data['average_rating'],
        show_data['num_viewers (mil)']
    )
    
    results.append({
        'show_name': show,
        'correlation': r,
        'p_value': p_value
    })

print(pd.DataFrame(results), "\n")

r, p_value = pearsonr(
    combined['average_rating'],
    combined['num_viewers (mil)']
)

print('All shows combined')
print('Correlation (r):', round(r, 3))
print('P-value:', round(p_value, 3))
```

It appears that none of these relationships are statistically significant, if we were to use the conventional 0.05 threshold. However, again, any p-value wouldn't be meaningful anyway in this situation. What's important is that the correlation coefficient indicates that the relationship between a show's rating and viewership within a season is quite strong. When we build our model to predict dating app usage/revenue, we need to make sure to not include both the same show's viewership data and its ratings data; we also need to be cautious when using viewership data from one show and ratings data from another show.

### 3.2. Standard OLS Regression

We first decided to try standard OLS Regression to see the immediate relationship between our independent and dependent variables (show popularity versus app popularity). OLS is a lightweight, simple test, and works well as a first step for determining simple relationships. We decided to make two models, one to predict dating app users and the other to predict dating app revenue. Both will have dating show viewers and ratings as predictors.  

Importantly, we don't intend for this step to be our main analysis. Previously found multicolinearity between dating show viewership and rating exaggerates any small changes in the data, causing inflated and inaccurate results. This analysis is mainly meant for initial introspection, which can hopefully inform our future analysis decisions. 

#### 3.2.1. OLS Regression: Dating Shows predicting Dating App Revenue

Our first model has our three dating shows' mean viewership and mean ratings for a given year as predictors for dating app revenue in the same year.

```python
viewership_agg = show_viewership.groupby('year')['num_viewers (mil)'].mean().reset_index()

ratings_agg = show_ratings.groupby('year of airing')['rating'].mean().reset_index()
ratings_agg.columns = ['year', 'rating_mean']

# Merge all on year
df = revenue_users.merge(viewership_agg, on='year', how='left')
df = df.merge(ratings_agg, on='year', how='left')
df['users (mm)'] = pd.to_numeric(df['users (mm)'], errors='coerce')
df = df.dropna(subset=['users (mm)', 'num_viewers (mil)', 'rating_mean'])

# OLS
revenue_model = smf.ols('Q("revenue (mm)") ~ Q("num_viewers (mil)") + rating_mean', data=df).fit()

print(revenue_model.summary())
```

#### 3.2.2. OLS Regression: Dating Shows predicting Dating App User Count

Our second model has the same predictors, except this time, it's meant to predict dating app user counts. 

```python
users_model = smf.ols('Q("users (mm)") ~ Q("num_viewers (mil)") + rating_mean', data=df).fit()

print(users_model.summary())
```

#### 3.2.3. Interim Discussion

Our results are highly insignificant, with all p-values being far above .05, and with the r^2 values being extremely low. Based on these results alone, we would fail to reject the null hypothesis, which is that neither dating show viewership nor ratings are correlated with dating app revenue and user count. 

However, as mentioned, simple OLS Regression is clearly not the ideal way to analyze our data. High Durbin-Watson values for both users and revenue (0.808 and 0.638) suggest strong positive autocorrelation. This is most likely due to the fact that our data (particularly for dating app revenue and user numbers) follows the progression of time, meaning that each data point's value is highly correlated with the years immediately prior to or after it. This is an issue for OLS because it assumes no autocorrelation between data points. Furthermore, as previously addressed, our dating show variables are multicollinear, causing issues with the accuracy of our results and violating initial OLS assumptions. So, we go on to transform our data in the next step below in order to address issues with skewness and autocorrelation, and to rearrange our models so that multicollinearity isn't an issue. 

### 3.3. OLS Regression with Log-Log Transformation and Newey-West Robust Standard Error Adjustment

As listed above, standard OLS Regression has its setbacks and does not work well with our data. Aside from issues with autocorrelation and multicollinearity, we also realized that we must check for skewness in the data. Although OLS regression does not have a strong requirement for normality, it does have issues with skewed data. 

#### 3.3.1. Assessing Need for Log Transformations

We first assess the need for a log transformation to account for the skewness of the data, which, if left unattended, compromises the reliability of p-values and confidence intervals due to the disproportionate weighting of outliers.

To examine whether we need log transformations, we decided to check for skewness in all our datasets:

```python
sns.histplot(data=show_viewership['num_viewers (mil)'], color='red',kde=True) # checking the skewness of show viewership dataset
plt.title("Viewership Data Before Log Transformation")
```

The show viewership dataset is clearly right-skewed upon observation and hence, requires a log transformation.

```python
show_viewership['log_viewers']=np.log1p(show_viewership['num_viewers (mil)'])
sns.histplot(data=show_viewership['log_viewers'], color='green',kde=True) 
plt.title("Viewership Data After Log Transformation")
```

The log-transformed show viewership dataset now shows a normal distribution suitable for OLS Regression

```python
sns.histplot(data=show_ratings['rating'], color='skyblue',kde=True) # checking the skewness of show ratings dataset
plt.title("Ratings Data Before Log Transformation")
```

The show ratings dataset is not skewed upon observation and, hence, does not require a log transformation. It has an approximately bimodal curve shape and is suitable for OLS Regression directly.

```python
sns.histplot(data=revenue_users['revenue (mm)'], color='purple',kde=True) # checking the skewness of dating app revenue dataset
plt.title("Revenue Data Before Log Transformation")
```

The dating apps revenue dataset seems to be right-skewed upon observation, and hence, may require a log transformation.

```python
revenue_users['log_revenue']=np.log1p(revenue_users['revenue (mm)'])
sns.histplot(data=revenue_users['log_revenue'], color='green',kde=True) 
plt.title("Revenue Data After Log Transformation")
```

The Log transformation does not make things better in this case. In fact, it makes it seem even more skewed, just towards the left this time. Hence, we will preserve the original data, even though it is imperfect. 

```python
sns.histplot(data=revenue_users['users (mm)'].dropna(), color='pink',kde=True) # checking the skewness of dating app users dataset
plt.title("User Data Before Log Transformation")
```

The dating apps users dataset is relatively normally distrbuted. We should not need any log transformations here. 

#### 3.3.2 Summary of Log-Log transformations
A log transformation helps transform right-skewed datasets into a normal distribution, linearize all non-linear relationships, stabilize variance to prevent heteroskedasticity and measure elasticity. Our dating show viewership data was heavily skewed, so we applied a log transformation to it. Dating show revenue data has a tolerable skewness, and log transformation actually overcompensates and makes it left skewed instead. The dating app users data remains left-skewed (and has not been transformed to prevent worsening of the skewness) and dating show ratings already has a normal distribution.

#### 3.3.3. Newey West Robust Standard Errors
Our data is a time-series spanning over the years 2019-2026 and this causes a major drawback for OLS Regression, which is autocorrelation (a variable being correated with itself over various instances of time). Standard OLS assumes every data point is independent but this is false since the error term in each year of our datasets is highly likely to be correlated with the standard error from the previous year. In order to fix this, we decided to apply Newey West Standard Error Adjustment which will help stabilize variances to prevent heteroskedasticity and account for autocorrelation as much as possible.

Applying OLS Regression with the aforementioned log transformation and Newey-West Robust Standard Error Adjustment will allow us to identify if there is a correlation between dating show viewers and ratings and dating app revenue or users. Importantly, as uncovered during our EDA, dating show viewership and ratings are correlated. Hence, we also should not use viewership and ratings simultaneously as predictors in the same model, especially if they come from the same show. 

To make future analysis easier, we will create a new combined data frame. It will have two columns for dating apps (total revenue across all apps, total users across Bumble and Hinge since Tinder is missing data here), and each row will be the data for each year. We care more about the differences between dating shows than we do for dating apps (see our background section), so we aggregate data for dating apps rather than dating shows. For dating shows, we will add two more columns to our combined data frame for each show (one for their viewership in one year with log transformation, and another for their ratings in one year), totaling six new columns in total. To fix the issue of multicollinearity, we will make sure that a viewership column and a rating column are never fitted into the same model. 

```python
# Create app-level aggregates from the original loaded data without changing the source tables
apps_agg = (
    revenue_users[
        revenue_users['app'].isin(['Bumble', 'Hinge'])
    ]
    .groupby('year', as_index=False)
    .agg({'revenue (mm)': 'sum', 'users (mm)': 'sum'})
    .rename(columns={
        'revenue (mm)': 'total_revenue_mm',
        'users (mm)': 'total_users_mm',
    })
)

show_viewership_by_year = (
    show_viewership
    .groupby(['year', 'show_name'], as_index=False)['log_viewers']
    .mean()
)
show_ratings_by_year = (
    show_ratings
    .rename(columns={'year of airing': 'year'})
    .groupby(['year', 'show_name'], as_index=False)['rating']
    .mean()
)
```

```python
viewership_wide = (
    show_viewership_by_year
    .pivot(index='year', columns='show_name', values='log_viewers')
    .rename(columns=lambda x: f"log_viewers_{x.replace(' ', '_').replace('-', '_')}" )
    .reset_index()
)
ratings_wide = (
    show_ratings_by_year
    .pivot(index='year', columns='show_name', values='rating')
    .rename(columns=lambda x: f"rating_{x.replace(' ', '_').replace('-', '_')}" )
    .reset_index()
)

final_dataset = (
    apps_agg
    .merge(viewership_wide, on='year', how='left')
    .merge(ratings_wide, on='year', how='left')
    .sort_values('year')
    .reset_index(drop=True)
)

final_dataset.head()
```

#### 3.3.4. Dating Apps Revenue Model

Here, we use each of our shows' viewership and ratings data as separate predictors. Since we have three shows, we have 2 * 3 = 6 separate models to fit. Each predictor corresponds to the six rightmost show-popularity columns in `final_dataset`, using `total_revenue_mm` as the outcome.

Formula for each model: `total_revenue_mm ~ predictor`

We then summarize the adjusted R-squared and p-value of the correlation for each model in a table. 

```python
revenue_predictors = list(final_dataset.columns[-6:])

revenue_adjusted_r2_results = []

for predictor in revenue_predictors:
    revenue_model = smf.ols(f'total_revenue_mm ~ {predictor}', data=final_dataset)
    fitted_revenue = revenue_model.fit(cov_type='HAC', cov_kwds={'maxlags': 1}) ## Applying the Newey-West Standard Error Adjustment
    revenue_adjusted_r2_results.append({
        'Predictor': predictor,
        'Adjusted R-squared': fitted_revenue.rsquared_adj,
        'P-value': fitted_revenue.pvalues[predictor],
    })

revenue_adjusted_r2_table = (
    pd.DataFrame(revenue_adjusted_r2_results)
    .sort_values('Adjusted R-squared', ascending=False)
    .reset_index(drop=True)
)

revenue_adjusted_r2_table
```

The Bachelor’s viewership and rating ended up with the highest R-squared value for dating app revenue, suggesting a positive correlation between The Bachelor’s popularity and dating app revenue (though we are limited by a small sample size). The R-squared value resulting from The Bachelor’s viewership model is especially high, raising mild suspicions about its validity. Given our data, we hypothesize that the issues would most likely be caused by small sample size, as there is only one data point for viewership per year. This is an inevitable limitation given our resources.
The Bachelor's viewership also ended up with the lowest p-value, indicating significant results. In fact, all shows except for Love Island USA result in a p-value lower than 0.05, suggesting significance and rejection of the null hypothesis.

To contrast the best model, we will also plot the predictor of the worst non-null model (ratings_Love_Island_USA) against the predicted value.

#### 3.3.6. Dating Apps Users Model

Here we fit another six separate OLS models, one for each of the six rightmost show-popularity columns in `final_dataset`, but this time, we use `total_users_mm` as the outcome. Again, we summarize the performance of each model in a table. 

Formula for each model: `total_users_mm ~ predictor`

```python
users_predictors = list(final_dataset.columns[-6:])

users_adjusted_r2_results = []

for predictor in users_predictors:
    users_model = smf.ols(f'total_users_mm ~ {predictor}', data=final_dataset)
    fitted_users = users_model.fit(cov_type='HAC', cov_kwds={'maxlags': 1}) ## Applying the Newey-West Standard Error Adjustment
    users_adjusted_r2_results.append({
        'Predictor': predictor,
        'Adjusted R-squared': fitted_users.rsquared_adj,
        'P-value': fitted_users.pvalues[predictor],
    })

users_adjusted_r2_table = (
    pd.DataFrame(users_adjusted_r2_results)
    .sort_values('Adjusted R-squared', ascending=False)
    .reset_index(drop=True)
)

users_adjusted_r2_table
```

The Bachelor’s viewership alongside dating app user count ended up with the highest adjusted R-squared value, implying positive correlation between the two factors. The R-squared value of Love is Blind viewership is at a very similar value. We note that The Bachelor has the highest R-squared in both dating app user and dating app revenue tables due to the previously mentioned reasons. It’s also interesting to note that several dating shows have quite different R-squared values when being predicted by dating app user counts versus revenue, despite these two factors being multicollinear.
Again, all of the shows except for Love Island USA have p-values lower than 0.05, indicating significance and the ability to reject the null hypothesis.

#### 3.3.8. Interpretations

We acknowledge that the model is limited by the small sample size (14 samples) as indicated by the warning and therefore, is insufficient to satisfy the assumptions made by OLS regression. A larger dataset is required to make a definitive claim, which is beyond the scope of this project. Overall, an increase in dating show viewership and ratings is correlated with an increase in dating app user counts and revenue, excluding Love Is Blind ratings, Love Island USA Ratings and log transformed Love Island USA viewers (insufficient data) compared to both user count and revenue. Again, our results should not be seen as definitive. We encourage further replications with larger and more complete datasets. 

One worry we had about our results was that, perhaps, dating app popularity and dating show popularity were heavily correlated because time was a confound. That is, perhaps changes in both dating app popularity and dating show popularity reflect the passage of time; the observed relationship between the two might be a result of this shared association with time, rather than any real association between the two. However, returning back to the EDA, although dating app revenue and usage do increase with time (which makes sense, given these metrics might naturally accumulate), the measures we use to operationalize dating show popularity do not. Therefore, time as a confound most likely does not explain our observed relationship. 

Additionally, it's common for models to have more than one predictor to increase predictive power. It could be possible for us to apply forward selection and make our models even better by adding more predictors. However, we decided this was unnecessary for answering our question. Our primary goal is not to build a model that is good at predicting dating app usage and revenue based on metrics on dating shows, but rather, to see if a relationship between dating apps and dating shows emerges in the first place. We also wish to compare the predictive power of metrics from different dating shows, which would be harder to do if the same model had data from multiple shows. Therefore, we decide to conclude our analysis here.

## 5. Discussion and Conclusion

Our project is motivated by prior research suggesting that dating apps and dating shows share similar sociocultural sentiments, such as the gamification of dating, gender stereotypes, the endorsement of unrealistic beauty standards, and the sexualization of women. To investigate if such an association exists, we tested if dating app popularity (operationalized as the user count and revenue of three major dating apps) was correlated with dating show popularity (operationalized as three shows' viewership data and ratings). Notably, the three shows we selected are relatively diverse in their premises. We hypothesized that we would find a generic correlation between dating app popularity and dating show popularity, and if this correlation was stronger for certain dating shows, then it may suggest that the dating show in question might have more sociocultural overlap with dating apps. 

We found our data on publicly accessible databases, such as IMDb for ratings and Netflix reports for certain shows' viewership data. However, after conducting an exploratory analysis and initial standard OLS regressions, we found problems relating to multicollinearity, autocorrelation, and skewness in our datasets. To solve these issues, before our main analysis, we segregated potentially correlating predictors into separate models, adjusted our models using a method related to Newey-West Robust Standard Errors, and applied log transformations to skewed variables, respectively. 

We fit twelve models in total for our main analysis. This is because we had six predictors (2 for each dating show, which were viewership and ratings) and two predicted values (dating app usage and revenue), resulting in twelve possible combinations. We find that most models end up finding statistically significant relationships between dating app popularity and dating show popularity, and this relationship is particularly strong when the dating show is The Bachelor. One possible explanation for The Bachelor doing the best is because its premise best matches that of dating apps. More specifically, in The Bachelor, one male cast member dates and chooses among multiple female contestants, which highlights the competitive nature of dating. This is parallel to how dating apps also allow users to choose among many potential partners, and users need to "compete" with each other for attention. 

Of course, major limitations exist. For the sake of transparency, we wish to clearly state that no one should take our results at face value. Particularly, one limitation we wish to highlight is our small sample size, which severely limits the statistical validity of any regression models we run. This was unavoidable, given that more detailed data was inaccessible, but readers of this project should be aware of this. Another limitation is that our operationalizations of dating app and dating show popularity might be imperfect. People can be influenced by the contents of dating shows or the culture surrounding dating apps without viewing shows on Netflix or paying for a Tinder membership, but our data would not account for these people. Additionally, our data is at risk of being biased, specifically our ratings data. This is because only the most dedicated fans tend to actively rate shows and episodes, and these fans might not be objective when it comes to evaluating the quality of a show. Furthermore, although we endorse such an interpretation, shared sociocultural sentiments between dating apps and shows might not sufficiently explain the correlations we observe. 

Future research can investigate the three-way relationship between sociocultural sentiments about dating, dating app usage, and the viewing of dating shows more carefully, perhaps using survey methods rather than looking at aggregated metrics like we have. Additionally, we were limited by our lack of funding and constrained time commitment to this course project. Other researchers who are better resourced can substantiate our analyses with a larger dataset. Qualitative research would also be helpful. Specifically, future research can focus on delineating the various sociocultural sentiments reflected by prominent dating shows and document their societal effects. Research like this would be capable of supporting (or challenging) our interpretations of our results. 

Finally, we do not believe our project has any major negative ethical implications; all data was aggregated and publicly available already. We have also already discussed potential imperfections in our proxies and modeling processes, and we encourage skepticism when reading our results. We hope that our project can encourage people to reflect more critically on the prevalence of dating shows and dating apps in modern culture, though we also acknowledge that some people may have legitimate reasons for appreciating or using them.
